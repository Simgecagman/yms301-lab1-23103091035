### 🐞 Kasıtlı Hata #1 — None ile == Karşılaştırması

Java'daki `null`'un Python karşılığı `None`'dır. Ama `None` ile karşılaştırırken `==` yerine
`is` kullanmak gerekir. Önce yanlışını deneyelim:

````python
sonuc = None
if sonuc == None:
    print("Sonuc yok (== ile)")
````

Bu kod aslında **çalışır** ve doğru sonucu verir — hata almazsınız! Peki sorun ne? Sorun şu:
bazı özel nesneler `__eq__` metodunu override edip `== None` ile garip davranabilir (örn. bazı
NumPy/Pandas nesneleri). `is None`, nesnenin **kimliğini** (bellekteki tekliğini) karşılaştırır ve
hiçbir zaman yanıltılamaz — `None` programda her zaman **tek bir** nesnedir.

````python
sonuc = None
if sonuc is None:   # PEP 8'in onerdigi, guvenli yol
    print("Sonuc yok (is ile)")
````

👉 **Çıkarım:** `None`, boş string (`""`) ya da `0` ile **aynı şey değildir** — `None` "burada
hiç değer yok" demektir. PEP 8 kuralı: `None` karşılaştırmasında her zaman `is`/`is not` kullanın.

`type()` kesin tipi verir; `isinstance()` ise kalıtımı da hesaba katar — pratikte tercih edilen odur:

````python
print(type(21) == int)        # calisir ama onerilmez
print(isinstance(21, int))     # onerilen yol
print(isinstance(True, int))   # 🤔 bool, int'in alt sinifidir - sonuc ne olur?
````

🤔 **Tartışalım:** Son satırın çıktısı neden `True`? Python'da `bool` aslında `int`'in bir alt
sınıfıdır (`True == 1`, `False == 0`) — Java'da `boolean` ve `int` tamamen ayrı tiplerdir, bu yüzden
bu durum deneyimli Java programcılarını genelde şaşırtır.

## 2. f-string ve Slicing (Dilimleme)

f-string'in formatlama gücünü hatırlayalım — `{değişken:format_kodu}`:

````python
pi = 3.14159265
print(f"Pi yaklasik {pi:.2f}")      # 2 ondalik basamak
print(f"Pi yaklasik {pi:10.2f}")    # 10 karakterlik alanda saga yasli
isim = "ayşe"
print(f"{isim.upper()}")            # f-string icinde metot cagirmak serbest
````

Java/C'de doğrudan karşılığı olmayan, tamamen Python'a özgü bir özellik: **slicing**.
`[başlangıç:bitiş:adım]` ile bir string'in (ya da listenin) herhangi bir parçasını alırsınız —
Java'daki `substring()` ve `StringBuilder.reverse()`'ün ikisinin de yerini tutar:

````python
metin = "Yazilim Muhendisligi"
print(metin[0:7])     # "Yazilim" - 0'dan 7'ye kadar (7 dahil degil!)
print(metin[-1])      # son karakter
print(metin[-11:])    # son 11 karakter
print(metin[::-1])    # TUM metni ters cevirir (adim = -1)
print(metin[::2])     # her 2 karakterden birini al
````

🤔 **Tartışalım:** `metin[0:7]` neden 8 karakter değil 7 karakter veriyor? (İpucu: Python'da
dilimleme aralıkları her zaman "bitiş hariç" çalışır — bu, `range()` fonksiyonuyla da tutarlıdır
ve haftaya döngülerde tekrar karşımıza çıkacak.)

## 3. Fonksiyonlar — Bu Haftanın Asıl Konusu

Java'da bir metot tanımına kıyasla Python çok daha az "tören" (ceremony) ister — erişim
belirteci yok, tip bildirimi opsiyonel. Ama **tip ipucu** ve **docstring** yazmak profesyonel
Python'da standarttır ve dönem projesi şartnamesi bunu zorunlu kılıyor:

````python
def ortalama_hesapla(sayilar: list) -> float:
    """Bir sayi listesinin aritmetik ortalamasini dondurur."""
    return sum(sayilar) / len(sayilar)

print(ortalama_hesapla([80, 90, 70]))
help(ortalama_hesapla)  # docstring'i goster
````

### Varsayılan Parametreler — Java'da Overload Gerektiren Bir Şey

Java'da aynı fonksiyonu farklı parametre sayısıyla çağırabilmek için **overload** (aynı isimle
birden fazla metot) yazmanız gerekir. Python'da tek bir tanım, varsayılan değerle aynı işi görür:

````python
def selamla(isim: str, resmi: bool = False) -> str:
    """Kullaniciya normal ya da resmi bicimde selam verir."""
    if resmi:
        return f"Sayın {isim}, hoş geldiniz."
    return f"Selam {isim}!"

print(selamla("Ayşe"))
print(selamla("Ayşe", resmi=True))
````

### 🐞 Kasıtlı Hata #2 — Mutable Default Argument Tuzağı

Slaytta anlatılan Python'un en ünlü tuzaklarından birini şimdi siz de üretin. Aşağıdaki fonksiyon,
bir listeye eleman ekleyip onu döndürüyor gibi görünüyor — varsayılan parametre olarak boş bir
liste `[]` veriyoruz:

````python
def listeye_ekle(eleman, liste=[]):
    liste.append(eleman)
    return liste

print(listeye_ekle(1))   # beklenti: [1]
print(listeye_ekle(2))   # beklenti: [2]  ... GERCEKTE NE OLUYOR?
````

**Gördüğünüz gibi ikinci çağrı `[2]` değil `[1, 2]` döndürdü!** Hata mesajı yok, sessiz bir
mantık hatası — ve bu yüzden daha tehlikeli. Sebep: `liste=[]` varsayılan değeri, fonksiyon
**tanımlandığı anda, BİR KEZ** oluşturulur; her çağrı aynı liste nesnesini paylaşır. Java'da
metot parametreleri her çağrıda yeniden oluşur, bu yüzden bu durum Java programcılarını (hatta
deneyimli olanları bile) şaşırtır.

**Doğru çözüm:** varsayılan değer olarak `None` kullanıp, fonksiyon içinde gerçek listeyi oluşturun:

````python
def listeye_ekle_dogru(eleman, liste=None):
    if liste is None:        # az once gordugumuz "is None" kalibi burada ise yariyor!
        liste = []
    liste.append(eleman)
    return liste

print(listeye_ekle_dogru(1))
print(listeye_ekle_dogru(2))  # artik [2] - beklendigi gibi
````

👉 **Çıkarım:** Varsayılan parametre olarak **asla** değiştirilebilir (mutable) bir nesne
(liste, sözlük, küme) kullanmayın — yerine `None` verip fonksiyon içinde oluşturun. Bu, az önce
öğrendiğimiz `is None` kalıbının neden bu kadar önemli olduğunu da gösteriyor.

### Birden Fazla Değer Döndürmek

Java/C'de bir fonksiyondan birden fazla değer döndürmek için bir dizi, nesne ya da struct
paketlemeniz gerekir. Python'da virgülle ayrılmış birden fazla değer doğrudan döndürülebilir
(perde arkasında bir **tuple** oluşturulur) ve doğrudan birden fazla değişkene bölünebilir:

````python
def min_max(sayilar: list) -> tuple:
    """Bir listenin en kucuk ve en buyuk degerini birlikte dondurur."""
    return min(sayilar), max(sayilar)

kucuk, buyuk = min_max([4, 1, 9, 2])
print(f"En kucuk: {kucuk}, en buyuk: {buyuk}")
````

## 4. 🎯 Mini Proje: Parola İstatistik Fonksiyonu v0.1

Geçen haftanın (varsayımsal) düz-script yaklaşımının aksine, bu sefer **tek bir fonksiyon**
yazacağız: tip ipuçlu, docstring'li, bir `dict` döndüren. Bu, dönem projesinin "modüler kod"
gereksiniminin en küçük ölçekte provasıdır.

*(Not: gerçek bir parolanızı girmeyin — bu sadece bir alıştırma.)*

````python
def parola_istatistikleri(parola: str) -> dict:
    """
    Bir parolanin temel istatistiklerini hesaplar.

    Args:
        parola: incelenecek parola metni.

    Returns:
        uzunluk, harf/rakam/buyuk harf icerip icermedigi gibi
        bilgileri tutan bir sozluk (dict).
    """
    return {
        "uzunluk": len(parola),
        "sadece_harf_mi": parola.isalpha(),
        "sadece_rakam_mi": parola.isdigit(),
        "buyuk_harf_var_mi": parola != parola.lower(),
        "bosluk_var_mi": " " in parola,
    }
````

Fonksiyonumuzu çağıralım ve dönen `dict`'i inceleyelim:

````python
parola = input("Bir deneme parolasi girin: ")
istatistik = parola_istatistikleri(parola)
print(istatistik)
print(type(istatistik))
````

`dict`'in tek tek alanlarına köşeli parantezle erişiriz — Java'daki `HashMap.get()`'in sözdizimsel karşılığı:

````python
print(f"Uzunluk: {istatistik['uzunluk']}")
print(f"Buyuk harf iceriyor mu? {istatistik['buyuk_harf_var_mi']}")
````

Şimdi hepsini tek bir **rapor** haline getirelim:

````python
print("=" * 30)
print("PAROLA İSTATİSTİK RAPORU")
print("=" * 30)
for anahtar, deger in istatistik.items():
    print(f"{anahtar}: {deger}")
````

👉 Dikkat ettiyseniz son hücrede `for anahtar, deger in istatistik.items():` yazdık — bir
`dict`'in üzerinde döngü kurmanın standart yolu budur. Döngüleri detaylı olarak haftaya
işleyeceğiz; bugün yalnızca bir fonksiyonun dönüş değerini "profesyonelce" rapor hâline
getirebileceğinizi gördünüz.

### Görev 1 — Isınma: Toplam Hesapla (5 dk, bireysel)

### Görev 2 — Rehberli: Özel Karakter Alanı (10 dk, ikili)
`parola_istatistikleri`'ne yeni bir alan ekleyin: `"ozel_karakter_var_mi"`. (Lab Rehberi, Görev 2)

````python
# Görev 2 - Özel Karakter Alanı
def parola_istatistikleri(parola: str) -> dict:
    """Parolanin temel istatistiklerini hesaplar (ozel karakter alani eklendi)."""
    return {
        "uzunluk": len(parola),
        "sadece_harf_mi": parola.isalpha(),
        "sadece_rakam_mi": parola.isdigit(),
        "buyuk_harf_var_mi": parola != parola.lower(),
        "bosluk_var_mi": " " in parola,
        # harf/rakam DISINDA en az bir karakter varsa True
        "ozel_karakter_var_mi": any(not k.isalnum() for k in parola),
    }

print(parola_istatistikleri("abc123"))
print(parola_istatistikleri("Abc!23"))
````

### Görev 3 — Rehberli: Varsayılan Parametre Ekleyin (10 dk, ikili)
`min_max`'a `virgul_ile_mi: bool = False` parametresi ekleyin. (Lab Rehberi, Görev 3)

````python
# Görev 3 - Varsayılan Parametre
def min_max(sayilar: list, virgul_ile_mi: bool = False) -> tuple | str:
    """En kucuk ve en buyuk degeri tuple ya da 'kucuk,buyuk' string'i olarak dondurur."""
    kucuk, buyuk = min(sayilar), max(sayilar)
    if virgul_ile_mi:
        return f"{kucuk},{buyuk}"
    return kucuk, buyuk

print(min_max([4, 1, 9, 2]))
print(min_max([4, 1, 9, 2], virgul_ile_mi=True))
````

### Görev 4 — Bağımsız: Slicing ile Maskeleme (10 dk, ikili)
`"python123"` → `"p*******3"` gibi bir maskeleme fonksiyonu yazın. (Lab Rehberi, Görev 4)

````python
# Görev 4 - Slicing ile Maskeleme
def parola_maskele(parola: str) -> str:
    """Ilk ve son karakteri gosterir, aradakileri * ile maskeler."""
    if len(parola) <= 2:
        return parola
    return parola[0] + "*" * len(parola[1:-1]) + parola[-1]

print(parola_maskele("python123"))
````

### Görev 5 — Bağımsız: Parola Tekrar Sayısı (10 dk, ikili)
`parola_tekrar_sayisi(parola: str) -> dict` yazın: en çok tekrar eden karakteri ve adedini
döndürsün. (Lab Rehberi, Görev 5)

````python
# Görev 5 - Parola Tekrar Sayısı
def parola_tekrar_sayisi(parola: str) -> dict:
    """Parolada en cok tekrar eden karakteri ve adedini dondurur."""
    sayac = {}
    for karakter in parola:
        sayac[karakter] = sayac.get(karakter, 0) + 1
    en_cok = max(sayac, key=sayac.get)
    return {"karakter": en_cok, "adet": sayac[en_cok]}

print(parola_tekrar_sayisi("mississippi"))
````

### Görev 6 — Hata Ayıklama: Program Çalışmıyor (7 dk, ikili)

````python
# Görev 6 - DÜZELTİLDİ
# Hata: if satirinin sonunda ":" eksikti -> SyntaxError (expected ':')
# Python kodu calistirmadan once okurken (parse) bu hatayi yakalar.
def parola_uzunluk_kontrolu(parola: str, min_uzunluk: int = 8) -> bool:
    """Parola en az min_uzunluk karakterse True dondurur."""
    if len(parola) >= min_uzunluk:
        return True
    return False

print(parola_uzunluk_kontrolu("abc"))
print(parola_uzunluk_kontrolu("abcdefgh"))
````

### Görev 7 — Hata Ayıklama: Yanlış Sonuç Veriyor (8 dk, ikili)
Bu hücre ÇALIŞIYOR ama yanlış sonuç veriyor — hata mesajı yok (sessiz mantık hatası). Bulun ve
düzeltin. (Lab Rehberi, Görev 7)

````python
# Görev 7 - DÜZELTİLDİ
# Hata: fonksiyon en_uzun yerine kelimeler[0] donduruyordu.
# Sozdizimi ve tipler dogru oldugu icin hata mesaji yok -> sessiz mantik hatasi.
def en_uzun_kelime(cumle: str) -> str:
    """Cumledeki en uzun kelimeyi dondurur."""
    kelimeler = cumle.split()
    en_uzun = ""
    for kelime in kelimeler:
        if len(kelime) > len(en_uzun):
            en_uzun = kelime
    return en_uzun

print(en_uzun_kelime("bu bir deneme cumlesidir"))
````
